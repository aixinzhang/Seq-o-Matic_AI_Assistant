```python
start = datetime.now()
now=datetime.now()
time_diff = (now - start).total_seconds()

while time_diff < 305:
    heater.write(b'TEMP\r\n')
    s=heater.readline().decode('utf-8').strip()
    with open(os.path.join("D:",'temp.txt'), 'a') as fp:
        fp.write(str(s)+ '\n')
    fp.close()
    time.sleep(2)
    now = datetime.now()
    time_diff = (now - start).total_seconds()
    print(time_diff)
```
